```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]

plt.figure(figsize=(12, 5))
plt.plot(days, temperature, marker="o", linestyle="--")
plt.title("Weekly Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.grid(True, alpha=0.3)

for x, y in zip(days, temperature):
    plt.text(x, y + 0.3, y, ha="center")

plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]

plt.figure(figsize=(10, 6))

plt.barh(languages, students)

plt.title("Students by Programming Language")
plt.xlabel("Number of Students")
plt.ylabel("Languages")

plt.xlim(0, 40)

plt.grid(axis="x", linestyle="--", alpha=0.5)

for i, value in enumerate(students):
    plt.text(value + 0.5, i, str(value), va="center")

plt.gca().invert_yaxis()

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]

fig = plt.figure(figsize=(10, 6))

plt.scatter(
    study_hours,
    scores,
    c=ages,
    cmap="viridis",
    marker="o",
    s=80
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.grid(True)

cbar = plt.colorbar()
cbar.set_label("Age")

max_index = scores.index(max(scores))

plt.annotate(
    "Highest score",
    (study_hours[max_index], scores[max_index]),
    xytext=(-60, -20),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

ax_inset = fig.add_axes([0.16, 0.55, 0.22, 0.22])

ax_inset.bar(study_hours, scores)

ax_inset.set_title("Scores", fontsize=8)

ax_inset.set_xticks([2, 4, 6, 8, 10])
ax_inset.set_yticks([0, 20, 40, 60, 80])

plt.show()
```
